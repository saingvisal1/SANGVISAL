## Topic 1-Control flow

### 1.When STATEMENT (Switch)

```kotlin
val day = 3
when (day){
    1-> println("Monday")
    2-> println("Tuesday")
    3-> println("Wednesday")
    4-> println("Thursday")
    5-> println("Friday")
    6-> println("Saturday")
    7-> println("Sunday")
    else -> println("Invalid Input")
}
```

### 2.When EXPRESSION

```kotlin
val day = 9
val displayName = when(day){
    1-> "Monday"
    2-> "Tuesday"
    3-> "Wednesday"
    4-> "Thursday"
    5-> "Friday"
    6-> "Saturday"
    7-> "Sunday"
    else -> "Input Invalid"
}
println(displayName)
```

### 3. When without argument

```kotlin
val tempurature = 28
val device = when{
    tempurature < 0 -> "Frezzing -stay inside"
    tempurature < 15 -> "Cold - bring a jacket"
    tempurature < 30 -> "Comfortable"
    else -> "Sunny day stay hydrate"
}
println(device)
```

### 4.Multiple values

```kotlin
val month = 14
val season = when(month){
    12,1,2 -> "Winter"
    3,4,5 -> "Spring"
    6,7,8 -> "Summer"
    9,10,11 -> " Autumn"
    else -> "Input Invalid"
}
println(season)
```

### 5.Range checks inside when

```kotlin
val score = 90
val grade = when(score){
    in 90..100 -> "A"
    in 75..89 -> "B"
    in 60..74 -> "C"
    else -> "F"
}
println(grade)
```

## Topic 3 - Function: Default Params, Named Argus, Single-Expression, Unit vs Return

## 1. Default Param

```kotlin
fun greet(name: String, greeting : String = "Hello" ){
    println("$greeting $name!")
}
greet("Dara")
greet("Panha","Hi")
```

### 2.Named Argument

```kotlin
fun createProfile(username: String, age: Int, isVIP: Boolean = false){
    println("$username $age is VIP Customer: $isVIP")
}
createProfile(username = "Visal",age = 18,isVIP = true)
```

### 3.Single-expression functions

```kotlin
//Traditional:
fun double(x: Int):Int{
    return x * 2
}
//single Expression
fun double(x:Int): Int = x * 2
fun square(x: Int) = x * 2
fun isEven(x: Int) = x % 2 == 0
```

### 4.Unit vs Return types

```kotlin

fun add(a: Int, b: Int) = println(a + b)
add(4,1)
```
